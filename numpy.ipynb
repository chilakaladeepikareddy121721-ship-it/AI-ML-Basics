Broadcasting – Definition
Broadcasting is a NumPy feature that allows arithmetic operations between arrays of different shapes.
It automatically expands the smaller array to match the shape of the larger array.
This expansion is virtual (no actual copying of data), making operations faster and memory-efficient.
Rules of Broadcasting (in Points)
Compare shapes from right to left.
NumPy starts comparing dimensions from the last (rightmost) dimension.
Dimensions must be equal or one of them must be 1.
If the dimensions are the same, broadcasting is possible.
If one dimension is 1, it is stretched to match the other dimension.
Missing dimensions are treated as 1.
If one array has fewer dimensions, NumPy assumes leading dimensions of size 1.
If dimensions are neither equal nor 1, broadcasting fails.
NumPy raises a ValueError.
The output array has the shape of the larger array after broadcasting.
