```python
import matplotlib.pyplot as plt
classes = ['python','R','AI','ML','DS']
class1_students = [30,60,70,80,20]
class2_students = [20,50,30,40,10]
class3_students = [10,40,20,20,5]
plt.figure(figsize = (16,9))
width = 0.4

plt.bar(classes,class1_students,width,color='b',label='class1_students')
plt.bar(classes ,class2_students,width,color='m',label='class2_students')
plt.bar(classes ,class3_students,width,color='y',label='class2_students')
plt.title('Bhar chat of MU class',fontsize=15)
plt.xlabel('Classes',fontsize=20)
plt.ylabel('Number of students',fontsize=20)
plt.show()
```

```python
import numpy as np
plt.figure(figsize = (16,9))
classes_index = np.arange(len(classes))
width = 0.2


plt.bar(classes_index,class1_students,width,color='b',label='class1_students')
plt.bar(classes_index + width ,class2_students,width,color='m',label='class2_students')
plt.bar(classes_index + width + width,class3_students,width,color='y',label='class2_students')
plt.xticks(classes_index + width,classes,rotation = 20)
plt.title('Bhar chat of MU class',fontsize=15)
plt.xlabel('Classes',fontsize=20)
plt.ylabel('Number of students',fontsize=20)
plt.show()
```

```python
ml_students_age = np.random.randint(18,28,(100))
py_students_age = np.random.randint(14,28,(100))
bins = [15,20,25,30,35,40,45]
plt.hist(ml_students_age,bins)
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.5)
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.5,histtype='step')
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.5,histtype='barstacked',align='left')
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.5,align='right')
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.9,orientation = 'horizontal')
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.9, orientation = 'horizontal',color='m')
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.9,color = 'g',label=(ml_students_age))

plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.9,label=(ml_students_age),orientation = 'horizontal')
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.legend()
plt.show()
```

```python
plt.hist(ml_students_age,bins,rwidth = 0.9,label=(ml_students_age))
plt.hist(py_students_age,bins,rwidth = 0.9,label=(ml_students_age))
plt.title('ml  students age histogram')
plt.xlabel('students age category')
plt.ylabel('no.of students age')
plt.legend()
plt.show()
```
