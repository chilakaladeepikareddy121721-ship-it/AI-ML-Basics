```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
globalwarming_df = pd.read_csv('/content/Who_is_responsible_for_global_warming (1).csv')
globalwarming_df
```

```python
globalwarming_df = globalwarming_df.drop(columns = ['Country Code','Indicator Name','Indicator Code'],axis=1)
globalwarming_df
```

```python
globalwarming_df=globalwarming_df.set_index('Country Name')
globalwarming_df
```

```python
sns.heatmap(globalwarming_df)
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(globalwarming_df)
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(globalwarming_df,vmin=0,vmax=21)
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(globalwarming_df,cmap='coolwarm')
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(globalwarming_df,cmap='coolwarm',annot=True)
```

```python
plt.figure(figsize=(16,9))
annot_kws={'fontsize' : 10,
           'fontstyle': 'italic',
           'color' : 'k',
           'alpha' : 0.6,
           'rotation' : 'vertical',
           'verticalalignment':'center',
           'backgroundcolor' : 'w'}
sns.heatmap(globalwarming_df,cmap='coolwarm',annot=True,annot_kws=annot_kws)
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(globalwarming_df,cmap='coolwarm',annot=True,linewidth=4,linecolor='k')
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(globalwarming_df,cbar=False)
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(globalwarming_df,cbar=False,xticklabels=False,yticklabels=False)
```
